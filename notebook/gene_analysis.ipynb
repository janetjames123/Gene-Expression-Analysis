# Bioinformatics Context:
# Each column represents a gene
# Each row represents a biological sample
# Values represent gene expression levels

```python
df_numeric = df.select_dtypes(include=['number'])
df_numeric.head()
```

```python
sns.heatmap(df_numeric.corr(), cmap="coolwarm")
plt.title("Gene Correlation Heatmap")
plt.show()
```

# Insight:
# Some genes show strong positive or negative correlation,
# suggesting possible co-regulation or shared biological pathways.


```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_numeric)

plt.scatter(pca_result[:,0], pca_result[:,1])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Gene Expression Data")
plt.show()
```

# Insight:
# PCA reduces high-dimensional gene data into 2 components,
# allowing visualization of sample clustering patterns.
